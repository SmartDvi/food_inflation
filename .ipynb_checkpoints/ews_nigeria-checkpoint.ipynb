# 🇳🇬 Food Price Shock EWS — Feature Engineering + ML
**DuckDB + Scikit-learn | Jupyter Lab**

```python
import duckdb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection   import RandomizedSearchCV, TimeSeriesSplit
from sklearn.ensemble          import RandomForestClassifier, GradientBoostingClassifier
from sklearn.preprocessing     import StandardScaler
from sklearn.metrics           import (
    classification_report, roc_curve, auc,
    f1_score, make_scorer, confusion_matrix, ConfusionMatrixDisplay
)

# ── Connect to your DuckDB file (change path as needed) ──
con = duckdb.connect('prices.duckdb')
print('✅ Connected to DuckDB')
```

---
## STEP 1 — Monthly Median Prices

```python
con.execute("""
    CREATE OR REPLACE VIEW monthly_prices AS
    SELECT
        DATE_TRUNC('month', date)::DATE  AS month,
        commodity,
        admin1,
        MEDIAN(price)                    AS price,
        COUNT(*)                         AS obs_count
    FROM prices
    GROUP BY 1, 2, 3
    ORDER BY commodity, admin1, month
""")

con.execute("SELECT * FROM monthly_prices LIMIT 5").df()
```

---
## STEP 2 — Lag Features

```python
con.execute("""
    CREATE OR REPLACE VIEW price_lags AS
    SELECT
        month, commodity, admin1, price,
        LAG(price, 1)  OVER w  AS price_lag_1m,
        LAG(price, 2)  OVER w  AS price_lag_2m,
        LAG(price, 3)  OVER w  AS price_lag_3m,
        LAG(price, 6)  OVER w  AS price_lag_6m,
        LAG(price, 12) OVER w  AS price_lag_12m
    FROM monthly_prices
    WINDOW w AS (
        PARTITION BY commodity, admin1
        ORDER BY month
    )
""")

con.execute("SELECT * FROM price_lags LIMIT 5").df()
```

---
## STEP 3 — Rolling Statistics (3m / 6m / 12m)

```python
con.execute("""
    CREATE OR REPLACE VIEW rolling_stats AS
    SELECT
        month, commodity, admin1, price,
        AVG(price)    OVER w3   AS roll_mean_3m,
        STDDEV(price) OVER w3   AS roll_std_3m,
        MAX(price)    OVER w3   AS roll_max_3m,
        MIN(price)    OVER w3   AS roll_min_3m,
        AVG(price)    OVER w6   AS roll_mean_6m,
        STDDEV(price) OVER w6   AS roll_std_6m,
        AVG(price)    OVER w12  AS roll_mean_12m,
        STDDEV(price) OVER w12  AS roll_std_12m
    FROM monthly_prices
    WINDOW
        w3  AS (PARTITION BY commodity, admin1 ORDER BY month
                ROWS BETWEEN 2  PRECEDING AND CURRENT ROW),
        w6  AS (PARTITION BY commodity, admin1 ORDER BY month
                ROWS BETWEEN 5  PRECEDING AND CURRENT ROW),
        w12 AS (PARTITION BY commodity, admin1 ORDER BY month
                ROWS BETWEEN 11 PRECEDING AND CURRENT ROW)
""")

con.execute("SELECT * FROM rolling_stats LIMIT 5").df()
```

---
## STEP 4 — Momentum & Ratio Features

```python
con.execute("""
    CREATE OR REPLACE VIEW momentum_features AS
    SELECT
        month, commodity, admin1, price,
        roll_mean_3m, roll_std_3m, roll_mean_12m, roll_std_12m,

        -- Z-score vs 12-month baseline
        CASE WHEN roll_std_12m > 0
             THEN ROUND((price - roll_mean_12m) / roll_std_12m, 4)
             ELSE NULL END                              AS zscore_12m,

        -- Price-to-rolling-mean ratios
        CASE WHEN roll_mean_3m  > 0
             THEN ROUND(price / roll_mean_3m,  4)
             ELSE NULL END                              AS price_to_mean_3m,

        CASE WHEN roll_mean_12m > 0
             THEN ROUND(price / roll_mean_12m, 4)
             ELSE NULL END                              AS price_to_mean_12m,

        -- Short vs long mean cross (trend signal)
        CASE WHEN roll_mean_12m > 0
             THEN ROUND(roll_mean_3m / roll_mean_12m, 4)
             ELSE NULL END                              AS mean_cross_ratio,

        -- Range position within 3m window
        CASE WHEN (roll_max_3m - roll_min_3m) > 0
             THEN ROUND((price - roll_min_3m) / (roll_max_3m - roll_min_3m), 4)
             ELSE NULL END                              AS range_position_3m
    FROM rolling_stats
""")

con.execute("SELECT * FROM momentum_features LIMIT 5").df()
```

---
## STEP 5 — Change Features (MoM, QoQ, YoY, Acceleration)

```python
con.execute("""
    CREATE OR REPLACE VIEW change_features AS
    SELECT
        p.month, p.commodity, p.admin1, p.price,

        CASE WHEN p.price_lag_1m  > 0
             THEN ROUND((p.price - p.price_lag_1m)  / p.price_lag_1m  * 100, 2)
             ELSE NULL END  AS mom_pct,

        CASE WHEN p.price_lag_3m  > 0
             THEN ROUND((p.price - p.price_lag_3m)  / p.price_lag_3m  * 100, 2)
             ELSE NULL END  AS qoq_pct,

        CASE WHEN p.price_lag_12m > 0
             THEN ROUND((p.price - p.price_lag_12m) / p.price_lag_12m * 100, 2)
             ELSE NULL END  AS yoy_pct,

        -- Acceleration: change in MoM rate
        CASE WHEN p.price_lag_1m > 0 AND p.price_lag_2m > 0
             THEN ROUND(
                    ((p.price        - p.price_lag_1m) / p.price_lag_1m)
                  - ((p.price_lag_1m - p.price_lag_2m) / p.price_lag_2m), 4)
             ELSE NULL END  AS mom_acceleration
    FROM price_lags p
""")

con.execute("SELECT * FROM change_features LIMIT 5").df()
```

---
## STEP 6 — Seasonal Features

```python
con.execute("""
    CREATE OR REPLACE VIEW seasonal_features AS
    SELECT
        month, commodity, admin1,
        MONTH(month)                                      AS month_num,
        QUARTER(month)                                    AS quarter,
        YEAR(month)                                       AS year_num,
        ROUND(SIN(2 * PI() * MONTH(month) / 12), 4)      AS month_sin,
        ROUND(COS(2 * PI() * MONTH(month) / 12), 4)      AS month_cos,
        -- Jun–Aug = lean season (historically high prices in N. Nigeria)
        CASE WHEN MONTH(month) BETWEEN 6  AND 8  THEN 1 ELSE 0 END AS lean_season_flag,
        -- Oct–Dec = harvest season (typically lower prices)
        CASE WHEN MONTH(month) BETWEEN 10 AND 12 THEN 1 ELSE 0 END AS harvest_flag
    FROM monthly_prices
""")

con.execute("SELECT * FROM seasonal_features LIMIT 5").df()
```

---
## STEP 7 — Shock Labels (Target Variable)

```python
con.execute("""
    CREATE OR REPLACE VIEW shock_labels AS
    WITH base AS (
        SELECT
            month, commodity, admin1, price,
            roll_mean_3m, roll_std_3m,
            CASE
                WHEN roll_std_3m > 0
                 AND price > roll_mean_3m + roll_std_3m THEN 1
                ELSE 0
            END AS shock_this_month
        FROM rolling_stats
    )
    SELECT
        month, commodity, admin1, price,
        shock_this_month,
        -- TARGET: shock in the NEXT month
        LEAD(shock_this_month, 1) OVER (
            PARTITION BY commodity, admin1
            ORDER BY month
        ) AS shock_next_month,
        -- How many shocks in last 3 months (severity signal)
        SUM(shock_this_month) OVER (
            PARTITION BY commodity, admin1
            ORDER BY month
            ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
        ) AS shocks_last_3m
    FROM base
""")

# Class balance check
con.execute("""
    SELECT
        shock_next_month                                        AS label,
        COUNT(*)                                                AS count,
        ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 1)     AS pct
    FROM shock_labels
    WHERE shock_next_month IS NOT NULL
    GROUP BY shock_next_month
    ORDER BY label
""").df()
```

---
## STEP 8 — Build Final ML Feature Table

```python
con.execute("""
    CREATE OR REPLACE TABLE ml_features AS
    SELECT
        sl.month,
        sl.commodity,
        sl.admin1,
        YEAR(sl.month)   AS year,
        MONTH(sl.month)  AS month_num,
        sl.price,
        pl.price_lag_1m, pl.price_lag_2m, pl.price_lag_3m,
        pl.price_lag_6m, pl.price_lag_12m,
        rs.roll_mean_3m,  rs.roll_std_3m,
        rs.roll_mean_6m,  rs.roll_std_6m,
        rs.roll_mean_12m, rs.roll_std_12m,
        mf.zscore_12m,
        mf.price_to_mean_3m, mf.price_to_mean_12m,
        mf.mean_cross_ratio, mf.range_position_3m,
        cf.mom_pct, cf.qoq_pct, cf.yoy_pct, cf.mom_acceleration,
        sf.month_sin, sf.month_cos,
        sf.lean_season_flag, sf.harvest_flag, sf.quarter,
        sl.shock_this_month, sl.shocks_last_3m,
        sl.shock_next_month                        AS target
    FROM shock_labels      sl
    JOIN price_lags        pl ON sl.month=pl.month AND sl.commodity=pl.commodity AND sl.admin1=pl.admin1
    JOIN rolling_stats     rs ON sl.month=rs.month AND sl.commodity=rs.commodity AND sl.admin1=rs.admin1
    JOIN momentum_features mf ON sl.month=mf.month AND sl.commodity=mf.commodity AND sl.admin1=mf.admin1
    JOIN change_features   cf ON sl.month=cf.month AND sl.commodity=cf.commodity AND sl.admin1=cf.admin1
    JOIN seasonal_features sf ON sl.month=sf.month AND sl.commodity=sf.commodity AND sl.admin1=sf.admin1
    WHERE sl.shock_next_month IS NOT NULL
      AND pl.price_lag_12m    IS NOT NULL
      AND rs.roll_std_12m     IS NOT NULL
    ORDER BY sl.commodity, sl.admin1, sl.month
""")

# Audit
con.execute("""
    SELECT
        COUNT(*)                  AS total_rows,
        COUNT(DISTINCT commodity) AS commodities,
        COUNT(DISTINCT admin1)    AS states,
        MIN(month)                AS earliest,
        MAX(month)                AS latest,
        ROUND(AVG(target)*100,1)  AS shock_rate_pct
    FROM ml_features
""").df()
```

---
## STEP 9 — Load Features into Pandas for ML

```python
FEATURE_COLS = [
    'price',
    'price_lag_1m','price_lag_2m','price_lag_3m',
    'price_lag_6m','price_lag_12m',
    'roll_mean_3m','roll_std_3m',
    'roll_mean_6m','roll_std_6m',
    'roll_mean_12m','roll_std_12m',
    'zscore_12m',
    'price_to_mean_3m','price_to_mean_12m',
    'mean_cross_ratio','range_position_3m',
    'mom_pct','qoq_pct','yoy_pct','mom_acceleration',
    'month_sin','month_cos',
    'lean_season_flag','harvest_flag','quarter',
    'shock_this_month','shocks_last_3m'
]
TARGET_COL   = 'target'
CUTOFF_YEAR  = 2022

df = con.execute("SELECT * FROM ml_features ORDER BY month").df()

train = df[df['year'] <  CUTOFF_YEAR]
test  = df[df['year'] >= CUTOFF_YEAR]

X_tr = train[FEATURE_COLS].values
y_tr = train[TARGET_COL].values.astype(int)
X_te = test[FEATURE_COLS].values
y_te = test[TARGET_COL].values.astype(int)

print(f'Train: {len(X_tr):,} rows | Shock rate: {y_tr.mean()*100:.1f}%')
print(f'Test : {len(X_te):,} rows | Shock rate: {y_te.mean()*100:.1f}%')
```

---
## STEP 10 — Scale Features

```python
scaler  = StandardScaler()
X_tr_s  = scaler.fit_transform(X_tr)
X_te_s  = scaler.transform(X_te)
print('✅ Scaling done')
```

---
## STEP 11 — Tune Random Forest

```python
tscv   = TimeSeriesSplit(n_splits=5)
scorer = make_scorer(f1_score, pos_label=1)

rf_params = {
    'n_estimators'     : [200, 300, 500],
    'max_depth'        : [4, 6, 8, 10, None],
    'min_samples_leaf' : [1, 2, 5, 10],
    'min_samples_split': [2, 5, 10],
    'max_features'     : ['sqrt', 'log2', 0.5],
    'class_weight'     : ['balanced', 'balanced_subsample',
                          {0:1,1:2}, {0:1,1:3}]
}

rf_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42),
    param_distributions=rf_params,
    n_iter=40, cv=tscv, scoring=scorer,
    n_jobs=-1, random_state=42, verbose=1
)
rf_search.fit(X_tr_s, y_tr)

print(f'\nBest RF params : {rf_search.best_params_}')
print(f'Best RF CV F1  : {rf_search.best_score_:.3f}')
```

---
## STEP 12 — Tune Gradient Boosting

```python
gb_params = {
    'n_estimators'    : [200, 300, 500],
    'max_depth'       : [2, 3, 4, 5],
    'learning_rate'   : [0.01, 0.03, 0.05, 0.1],
    'subsample'       : [0.6, 0.7, 0.8, 1.0],
    'min_samples_leaf': [1, 2, 5, 10],
    'max_features'    : ['sqrt', 'log2', None]
}

gb_search = RandomizedSearchCV(
    GradientBoostingClassifier(random_state=42),
    param_distributions=gb_params,
    n_iter=40, cv=tscv, scoring=scorer,
    n_jobs=-1, random_state=42, verbose=1
)
gb_search.fit(X_tr_s, y_tr)

print(f'\nBest GBM params : {gb_search.best_params_}')
print(f'Best GBM CV F1  : {gb_search.best_score_:.3f}')
```

---
## STEP 13 — Ensemble Evaluation

```python
rf_best  = rf_search.best_estimator_
gb_best  = gb_search.best_estimator_

rf_prob  = rf_best.predict_proba(X_te_s)[:,1]
gb_prob  = gb_best.predict_proba(X_te_s)[:,1]
ens_prob = (rf_prob + gb_prob) / 2
ens_pred = (ens_prob >= 0.5).astype(int)

print(classification_report(
    y_te, ens_pred,
    target_names=['No Shock next month','Shock next month']
))

fpr, tpr, _ = roc_curve(y_te, ens_prob)
roc_auc = auc(fpr, tpr)
print(f'Ensemble AUC : {roc_auc:.3f}')
```

---
## STEP 14 — Threshold Tuning

```python
best_t, best_f1 = 0.5, 0.0
for t in np.arange(0.2, 0.8, 0.02):
    pred = (ens_prob >= t).astype(int)
    f1   = f1_score(y_te, pred, pos_label=1, zero_division=0)
    if f1 > best_f1:
        best_f1, best_t = f1, t

print(f'Best threshold : {best_t:.2f}  |  Shock F1 : {best_f1:.3f}')
final_pred = (ens_prob >= best_t).astype(int)
print(classification_report(
    y_te, final_pred,
    target_names=['No Shock next month','Shock next month']
))
```

---
## STEP 15 — Evaluation Plots

```python
fi = pd.DataFrame({
    'feature'       : FEATURE_COLS,
    'rf_importance' : rf_best.feature_importances_,
    'gbm_importance': gb_best.feature_importances_,
}).assign(
    avg_importance=lambda d: (d.rf_importance + d.gbm_importance) / 2
).sort_values('avg_importance', ascending=False).reset_index(drop=True)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle(
    'EWS Ensemble — RF + GBM | Food Price Shock Prediction (Nigeria)',
    fontsize=13, fontweight='bold'
)

# ROC
axes[0].plot(fpr, tpr, color='#1565C0', lw=2,
             label=f'Ensemble AUC = {roc_auc:.3f}')
axes[0].plot([0,1],[0,1],'k--',lw=1)
axes[0].set(title='ROC Curve', xlabel='FPR', ylabel='TPR')
axes[0].legend(loc='lower right')

# Confusion matrix
cm = confusion_matrix(y_te, final_pred)
ConfusionMatrixDisplay(cm,
    display_labels=['No Shock','Shock']).plot(
    ax=axes[1], colorbar=False, cmap='Blues')
axes[1].set_title(f'Confusion Matrix (t={best_t:.2f})')

# Feature importance
top15 = fi.head(15)
axes[2].barh(top15['feature'][::-1], top15['avg_importance'][::-1],
             color='#E65100')
axes[2].set(title='Top 15 Features', xlabel='Avg Importance')

plt.tight_layout()
plt.savefig('ews_evaluation.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## STEP 16 — Write Predictions back to DuckDB

```python
# Score the full dataset
X_all_s        = scaler.transform(df[FEATURE_COLS].values)
df['rf_prob']  = rf_best.predict_proba(X_all_s)[:,1]
df['gbm_prob'] = gb_best.predict_proba(X_all_s)[:,1]
df['ens_prob'] = (df['rf_prob'] + df['gbm_prob']) / 2
df['shock_pred'] = (df['ens_prob'] >= best_t).astype(int)

# Register dataframe and write back
con.execute("DROP TABLE IF EXISTS ews_predictions")
con.execute("""
    CREATE TABLE ews_predictions AS
    SELECT
        month, commodity, admin1, year,
        price,
        target         AS actual_shock,
        rf_prob, gbm_prob, ens_prob, shock_pred
    FROM df
    ORDER BY commodity, admin1, month
""")

# Confirm
con.execute("""
    SELECT
        shock_pred,
        COUNT(*)                                               AS count,
        ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 1)    AS pct,
        ROUND(AVG(ens_prob), 3)                                AS avg_prob
    FROM ews_predictions
    GROUP BY shock_pred
    ORDER BY shock_pred
""").df()
```

---
## STEP 17 — Highest Risk States Right Now

```python
con.execute("""
    SELECT
        admin1,
        commodity,
        ROUND(AVG(ens_prob) * 100, 1)   AS avg_shock_prob_pct,
        SUM(shock_pred)                 AS months_flagged,
        COUNT(*)                        AS total_months
    FROM ews_predictions
    WHERE year >= 2023
    GROUP BY admin1, commodity
    ORDER BY avg_shock_prob_pct DESC
    LIMIT 15
""").df()
```
